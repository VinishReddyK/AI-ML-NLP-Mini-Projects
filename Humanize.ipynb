```python
import spacy
import random
import pyperclip

# Load the English NLP model
nlp = spacy.load('en_core_web_sm')

# Replacement dictionary for formal to casual words
replacement_dict = {
    'utilize': 'use',
    'numerous': 'many',
    'facilitate': 'help',
    'subsequently': 'later',
    'endeavor': 'try'
}

# List of stop words that can be removed except for some exceptions
stop_words_exceptions = {'and', 'of', 'is', 'to', 'on'}
stop_words_to_remove = {sw for sw in nlp.Defaults.stop_words if sw not in stop_words_exceptions}

def humanize_content(text):
    # Process the text using spaCy
    doc = nlp(text)

    humanized_text = []

    for sentence in doc.sents:
        words = []
        for token in sentence:
            word = token.text

            # Randomly replace with base form
            if random.random() < 0.2:
                word = token.lemma_

            # Replace formal words with casual ones
            word = replacement_dict.get(word.lower(), word)

            # Randomly remove stop words, with some exceptions
            if token.is_stop and word in stop_words_to_remove and random.random() < 0.1:
                continue

            words.append(word)

        # Join words into a sentence, apply random spacing
        sentence_str = ' '.join(words)
        sentence_str = sentence_str.replace(' .', '.')  # Ensure no space before dots

        # Randomly add extra spaces between sentences
        if random.random() < 0.5:
            sentence_str += '  '

        # Randomly remove punctuations except dots
        if random.random() < 0.7:
            sentence_str = ''.join([ch if ch == '.' or not ch in ',;:!?' else ' ' for ch in sentence_str])

        humanized_text.append(sentence_str.strip())

    return ' '.join(humanized_text)
```

```python
# Take input from user
user_input = input("Enter your text to be humanized: ")

# Humanize the content
humanized_text = humanize_content(user_input)

# Copy the humanized text to the clipboard
# pyperclip.copy(humanized_text)

# Output the humanized text
print(humanized_text)
print()
```
