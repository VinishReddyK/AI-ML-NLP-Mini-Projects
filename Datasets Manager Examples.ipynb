```python
import pandas as pd
```

Price Per liter vs cases data

```python
covid_df = pd.read_csv('/content/owid-covid-data.csv')

oil_df = pd.read_csv('/content/Petrol Dataset June 20 2022.csv', encoding='latin-1')

specific_date = '2020-06-20'

covid_df_specific_date = covid_df[covid_df['date'] == specific_date]

covid_df_specific_date = covid_df_specific_date.rename(columns={'location': 'country'})

covid_relevant = covid_df_specific_date[['country', 'total_cases_per_million']]

oil_relevant = oil_df[['Country', 'Price Per Liter (USD)']].rename(columns={'Country': 'country',
'Price Per Liter (USD)': 'price_per_liter'})

combined_df = pd.merge(oil_relevant, covid_relevant, on='country', how='inner')

combined_df.to_csv('Price Per liter vs cases data.csv', index=False)
```

Pollution vs Deaths

```python
covid_df = pd.read_csv('/content/owid-covid-data.csv')

pollu_df = pd.read_csv('/content/global air pollution dataset.csv')
pollu_df = pollu_df.drop_duplicates(subset=['Country'])

specific_date = '2020-06-20'

covid_df_specific_date = covid_df[covid_df['date'] == specific_date]

covid_df_specific_date = covid_df_specific_date.rename(columns={'location': 'country'})

covid_relevant = covid_df_specific_date[['country', 'total_deaths_per_million']]

oil_relevant = pollu_df[['Country', 'AQI Value']].rename(columns={'Country': 'country',
                                                                  'AQI Value': 'AQI'})

combined_df = pd.merge(oil_relevant, covid_relevant, on='country', how='inner')
combined_df = combined_df.dropna()

combined_df.to_csv('Pollution vs Deaths.csv', index=False)
```
