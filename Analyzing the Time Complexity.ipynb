```python
import time
import matplotlib.pyplot as plt

def FibRec(n):
  if n == 0 or n == 1:
    return n
  else:
    return FibRec(n-1) + FibRec(n-2)

def FibItr(n):
  if n == 0 or n == 1:
    return n
  else:
    a, b = 0, 1
    for i in range(2, n + 1):
      c = a + b
      a = b
      b = c
    return b

def SumCubesRec(n):
  if n == 0:
    return 0
  else:
    return n**3 + SumCubesRec(n-1)

def SumCubesItr(n):
  total = 0
  for i in range(n+1):
    total += i**3
  return total

def TimeMeassure(func, *args):
    start_time = time.process_time()
    func(*args)
    end_time = time.process_time()
    return end_time - start_time

inputs = list(range(1, 41))

def ItrTime(func):
  result = []
  for n in inputs:
    result.append(TimeMeassure(func, n))
  return result

FibRecTimes = ItrTime(FibRec)
FibItrTimes = ItrTime(FibItr)
SumCubesRecTimes = ItrTime(SumCubesRec)
SumCubesItrTimes = ItrTime(SumCubesItr)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(inputs, FibRecTimes, label='Fibonacci-Recursive')
plt.plot(inputs, FibItrTimes, label='Fibonacci-Iterative')
plt.plot(inputs, SumCubesRecTimes, label='Sum of Cubes-Recursive')
plt.plot(inputs, SumCubesItrTimes, label='Sum of Cubes-Iterative')
plt.xlabel('Input Size')
plt.ylabel('Processing Time-seconds')
plt.legend()
plt.title('Processing Time for Different Scenarios')
plt.show()
```
